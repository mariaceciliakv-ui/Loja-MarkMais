```python
#Loja MarkMais
#Parte 1- Listas
print("======== Bem-vindo à Loja MarkMais! ========")
produtos = ["Camiseta Básica", "Calça Jeans", "Jaqueta Corta-Vento"]
precos = [49.90, 129.90, 219.90]

# Lista os produtos (usando variáveis no singular para não sobrescrever as listas)
for produto, preco in zip(produtos, precos):
    print(f"O produto {produto} custa R${preco:.2f}")
```

```python
# Parte 2: Cadastrando Novos Itens
novo_produto = input("Digite o nome da nova peça de roupa: ")
novo_preco = float(input("Digite o preço da nova peça: "))

# Inserindo os valores
produtos.append(novo_produto)
precos.append(novo_preco)

# Mostrando o estoque
print("\n--- Estoque Atualizado ---")
for produto, preco in zip(produtos, precos):
    print(f"O produto {produto} custa R${preco:.2f}")
```

```python
# Parte 3: Removendo o Último Item (pop)
print("\n--- Removendo o Último Item ---")
if len(produtos) > 0:

# O .pop() remove e retorna o último elemento, guardando-o nas variáveis
    produto_removido = produtos.pop()
    preco_removido = precos.pop()
    print(f"Produto removido: {produto_removido} (R$ {preco_removido:.2f})")
else:
    print("Estoque vazio!")

# Mostrando o estoque final atualizado
print("\n--- Estoque Final Atualizado ---")
for produto, preco in zip(produtos, precos):
    print(f"O produto {produto} custa R${preco:.2f}")
```

```python
# Parte 4: Busca e Remoção por Nome
print("\n--- Busca e Remoção por Nome ---")
item_remover = input("Digite o nome exato da peça que deseja remover: ")

if item_remover in produtos:
    # Descobre a posição (índice) do item na lista de produtos
    i = produtos.index(item_remover)

    # Remove o produto e o preço correspondente usando o índice exato
    prod_rem = produtos.pop(i)
    preco_rem = precos.pop(i)
    print(f"Sucesso! O item '{prod_rem}' (R$ {preco_rem:.2f}) foi removido.")
else:
    print("Não existe!")

# Estoque Final Definitivo
print("\n--- Estoque Final Completo ---")
for produto, preco in zip(produtos, precos):
    print(f"O produto {produto} custa R${preco:.2f}")
```
